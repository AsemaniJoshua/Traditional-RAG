```python
### loading all types of files in a data storage directory instead of doing it one by one or group with text similarities

from langchain_community.document_loaders import PyMuPDFLoader
from importlib.resources import path
### Directory loading
from langchain_community.document_loaders import DirectoryLoader


# Loading all txt files in the data folder
txt_loader = DirectoryLoader(
    "../data", # parent folder
    glob="**/*.txt", # where to locate the fies and patterns for different extension
    loader_cls= TextLoader, # there can be a lsit of loaders
    # loader_kwargs={ "encoding": "utf-8" }, #this wont work for pdfloaders
    show_progress=True # needs tqdm installed
)

txt_documents = txt_loader.load()


# Loading all pdf files in the data folder
pdf_loader = DirectoryLoader(
    "../data", # parent folder
    glob="**/*.pdf", # where to locate the fies and patterns for different extension
    loader_cls= PyMuPDFLoader, # there can be a lsit of loaders
    # loader_kwargs={ "encoding": "utf-8" }, #this wont work for pdfloaders
    show_progress=True # needs tqdm installed
)

pdf_documents = pdf_loader.load()


# Combining all documents as one
documents = txt_documents + pdf_documents

print(documents)

```

```python
# loading from excel files from directory
import pandas as pd
from langchain_community.document_loaders import DataFrameLoader
from langchain_core.document_loaders import BaseLoader


# 1. Custom loader inheriting from BaseLoader
class custom_excel_loader(BaseLoader):
    def __init__(self, file_path: str):
        self.file_path = file_path

        
    def load(self):
        df = pd.read_excel(self.file_path, sheet_name=None)
        all_docs = []
        for sheet_name, sheet_table in df.items():
            # skip empty sheets
            if sheet_table.empty or len(sheet_table.columns) == 0:
                continue
            # merging all columns into page_content
            sheet_table["all_content"] = sheet_table.apply(
                lambda row: "\n".join([f"{col}: {val}" for col, val in row.items() if pd.notna(val)]),
                axis=1
            )
            # load this sheet using DataFrameLoader
            sheet_loader = DataFrameLoader(sheet_table, page_content_column="all_content")
            sheet_docs = sheet_loader.load()
            # attach file source and sheet name to metadata
            for doc in sheet_docs:
                doc.metadata["source"] = self.file_path
                doc.metadata["sheet_name"] = sheet_name
            all_docs.extend(sheet_docs)
        return all_docs



# Loading all pdf files in the data folder
excel_loader = DirectoryLoader(
    "../data", # parent folder
    glob="**/*.xlsx", # where to locate the fies and patterns for different extension
    loader_cls= custom_excel_loader, # there can be a lsit of loaders
    # loader_kwargs={ "encoding": "utf-8" }, #this wont work for pdfloaders
    show_progress=True # needs tqdm installed
)

excel_documents = excel_loader.load()

print(excel_documents)
```
